Why treat outliers before scaling?
1. Scaling is sensitive to outliers
Methods like StandardScaler (which uses mean and standard deviation) are strongly influenced by outliers.

If you scale first, outliers can distort the mean and standard deviation, leading to incorrect scaling for all other values.

2. Outliers can dominate scaled features
For example, if a column has values mostly in the 10–20 range but one value is 500, after scaling, most of the values become very small (e.g., −0.1 to 0.1), and the outlier becomes a large value (e.g., 5 or −5). This gives the outlier disproportionate influence on models like logistic regression or k-NN.

##### a.1.1.1 Python code to handle missing values

- `df.info()`  
  ➤ Gives quick information about the dataset.

- `df.isnull().sum()`  
  ➤ Count of null values per column.

- `df.dropna(axis=1)`  
  ➤ Drops columns that have null values. Use this when there are too many missing values in a column.

- `df.dropna(axis=0)`  
  ➤ Drops rows that have null values. Use this when you want to keep the column but remove incomplete rows.

- `df['property'] = df['property'].fillna(df['property'].mean())`  
  ➤ Fills NaN values with the mean of the column and modifies the original column.

- `df['property'] = df['property'].bfill()`  
  ➤ Backward fills null values using the next valid value in the column.

- `df['property'] = df['property'].ffill()`  
  ➤ Forward fills null values using the preceeding valid value in the column.
