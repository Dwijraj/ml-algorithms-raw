##### a.1.1.1 Python code to handle missing values

- `df.info()`  
  ➤ Gives quick information about the dataset.

- `df.isnull().sum()`  
  ➤ Count of null values per column.

- `df.dropna(axis=1)`  
  ➤ Drops columns that have null values. Use this when there are too many missing values in a column.

- `df.dropna(axis=0)`  
  ➤ Drops rows that have null values. Use this when you want to keep the column but remove incomplete rows.

- `df['property'] = df['property'].fillna(df['property'].mean())`  
  ➤ Fills NaN values with the mean of the column and modifies the original column.

- `df['property'] = df['property'].bfill()`  
  ➤ Backward fills null values using the next valid value in the column.

- `df['property'] = df['property'].ffill()`  
  ➤ Forward fills null values using the preceeding valid value in the column.
